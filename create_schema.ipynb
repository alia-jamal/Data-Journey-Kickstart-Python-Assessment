Create the database and connect it

```python
import sqlite3

# Create/connect to database
conn = sqlite3.connect("drilling_operations.db")

conn.execute("PRAGMA foreign_keys = ON;")
```

Create cursor (Execute SQL statement or feth query result)

```python
# Create cursor
cursor = conn.cursor()
```

Create REGIONS table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS REGIONS (
RegionId INTEGER PRIMARY KEY AUTOINCREMENT,
RegionName TEXT NOT NULL
)
""")
```

Create FIELDS table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS FIELDS (
FieldId INTEGER PRIMARY KEY AUTOINCREMENT,
RegionId INTEGER NOT NULL,
FieldName TEXT NOT NULL,

FOREIGN KEY (RegionId) REFERENCES REGIONS(RegionId)
)
""")
```

Create WELLS table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS WELLS (
WellId INTEGER PRIMARY KEY AUTOINCREMENT,
FieldId INTEGER NOT NULL,
PacId INTEGER NOT NULL,
WellName TEXT,
WellType TEXT,
WaterDepth REAL,
Year INTEGER,

FOREIGN KEY (FieldId) REFERENCES FIELDS(FieldId),
FOREIGN KEY (PacId) REFERENCES PAC(PacId)

)
""")
```

Create PAC table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS PAC (
PacId INTEGER PRIMARY KEY AUTOINCREMENT,
PacName TEXT NOT NULL
)
""")
```

Create RIGS table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS RIGS (
RigId INTEGER PRIMARY KEY AUTOINCREMENT,
RigName TEXT,
RigType TEXT
)
""")
```

Create WELL_OPERATIONS table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS WELL_OPERATIONS (
OperationId INTEGER PRIMARY KEY AUTOINCREMENT,
WellId INTEGER NOT NULL,
RigId INTEGER NOT NULL,
SpudDate DATETIME,
WellStartDateTime DATETIME,
WellEndDateTime DATETIME,
AfeCost REAL,
AfeDays REAL,
FinalCost REAL,
FinalDays REAL,
WellNptPercentage REAL,
WellNptPercentageWow REAL,
CompletionCostPlan REAL,
CompletionCostActual REAL,
DrillingPlanWcpf REAL,
DrillingActualWcpf REAL,

FOREIGN KEY (WellId) REFERENCES WELLS(WellId),
FOREIGN KEY (RigId) REFERENCES RIGS(RigId)

)
""")
```

Create REPORTS table

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS REPORTS (
ReportId INTEGER PRIMARY KEY AUTOINCREMENT,
OperationId INTEGER NOT NULL,
ReportType TEXT,
DocumentName TEXT,
DocumentDate DATETIME,
SubmittedAt DATETIME,
SubmittedBy TEXT,

FOREIGN KEY (OperationId) REFERENCES WELL_OPERATIONS(OperationId)

)
""")
```

Save the script

```python
# Save changes
conn.commit()
```

Close connection

```python
conn.close()
print("Database schema created successfully!")
```
