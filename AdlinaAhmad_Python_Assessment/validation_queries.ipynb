```python
######### SAMPLE QUERIES TO SEE LOADED DATA IN DB #################

conn = sqlite3.connect("drilling_operations.db")

tables = pd.read_sql("""
SELECT name 
FROM sqlite_master 
WHERE type='table'
""", conn)

results = []

for table in tables['name']:
    count = pd.read_sql(f"SELECT COUNT(*) AS row_count FROM {table}", conn)
    results.append({
        "table_name": table,
        "row_count": count['row_count'][0]
    })

summary_df = pd.DataFrame(results)

conn.close()

summary_df
```
