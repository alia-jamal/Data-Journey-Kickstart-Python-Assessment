1. Standardize string value

```python
import numpy as np

wellname_mapping = {
    'AAZ 1': 'AAZ-1',
    'BERGADING-C1 (A)': 'BERGADING-C1',
    'BUNGA KESUMBA A-4': 'BUNGA KESUMBA-A4',
    'LANG LEBAH - 2': 'LANG LEBAH-2',
    'TABU-B19 ST1': 'TABU-B19ST1',
    'TABU-B11 ST1': 'TABU-B11ST1',
    'TABU-B23 ST1': 'TABU-B23ST1',
    'KCP-1': 'KECAPI-1'
}

data['WellName'] = data['WellName'].replace(wellname_mapping)
```

Split the combine well name

```python
mask = data['WellName'] == 'WPPB-15/15ST1'

data.loc[mask, 'WellName'] = 'WPPB-15'

duplicated_rows = data.loc[mask].copy()
duplicated_rows['WellName'] = 'WPPB-15ST1'

data = pd.concat([data, duplicated_rows], ignore_index=True)
```

```python
print(data['WellName'].unique())
```

```python
#Check total rows and columns
num_rows = data.shape[0]
num_cols = data.shape[1]

print(f"Rows: {num_rows}")
print(f"Columns: {num_cols}")
```

```python
import re

def clean_fieldname(x):
    if pd.isna(x):
        return x

    x = x.upper().strip()

    # Remove known suffix noise
    x = re.sub(r'\b(DEV|EXP|EXPL|DEPS)\b.*$', '', x).strip()

    # Remove patterns like -A4, -1, A-111
    x = re.sub(r'[-][A-Z]?\d+$', '', x).strip()

    # Remove trailing single letters
    x = re.sub(r'\s+[A-Z]$', '', x).strip()

    # Standardize field names
    field_mapping = {
        'KCP': 'KECAPI',
        'KECAPI': 'KECAPI',
        'GAMUSUT KAKAP': 'GUMUSUT-KAKAP'
    }

    x = field_mapping.get(x, x)

    # Clean double spaces
    x = re.sub(r'\s+', ' ', x)

    return x

data['FieldName'] = data['FieldName'].apply(clean_fieldname)
```

```python
print(data['FieldName'].unique())
```

Fill in null value in RegionName

```python
data.loc[
    (data['FieldName'] == 'BULOH') & (data['RegionName'].isna()),
    'RegionName'
] = 'SB'   # replace with the correct region
```

```python
#Standardize same meaning with different words (missing special character, wrong spelling)
data['RigName'] = data['RigName'].str.strip()
data['RigName'] = data['RigName'].replace('NAGA - 5', 'NAGA-5')
data['RigName'] = data['RigName'].replace('NAGA 4', 'NAGA-4')
data['RigName'] = data['RigName'].replace('NAGA 7', 'NAGA-7')

data['RigType'] = data['RigType'].str.strip()
data['RigType'] = data['RigType'].replace('JACK UP', 'JACK-UP')
data['RigType'] = data['RigType'].replace('SEMI-SUB', 'SEMI-SUBMERSIBLE')
#data['RigType'] = data['RigType'].replace('Semi-Submersible', 'SEMI-SUBMERSIBLE')
data['RigType'] = data['RigType'].str.upper()

data['SubmittedBy'] = data['SubmittedBy'].str.strip()
data.loc[~data['SubmittedBy'].isin(['system', 'aem']), 'SubmittedBy'] = np.nan
```

2. Standardize decimal values

```python
#Convert to numeric datatype
data['WaterDepth'] = pd.to_numeric(data['WaterDepth'], errors='coerce')
data['AfeCost'] = pd.to_numeric(data['AfeCost'], errors='coerce')
data['AfeDays'] = pd.to_numeric(data['AfeDays'], errors='coerce')
data['FinalCost'] = pd.to_numeric(data['FinalCost'], errors='coerce')
data['FinalDays'] = pd.to_numeric(data['FinalDays'], errors='coerce')
data['WellNptPercentageWow'] = pd.to_numeric(data['WellNptPercentageWow'], errors='coerce')
data['WellNptPercentage'] = pd.to_numeric(data['WellNptPercentage'], errors='coerce')
data['CompletionCostPlan'] = pd.to_numeric(data['CompletionCostPlan'], errors='coerce')
data['CompletionCostActual'] = pd.to_numeric(data['CompletionCostActual'], errors='coerce')
data['DrillingPlanWcpf'] = pd.to_numeric(data['DrillingPlanWcpf'], errors='coerce')
data['DrillingActualWcpf'] = pd.to_numeric(data['DrillingActualWcpf'], errors='coerce')

#round to fixed number of decimal places
data['WaterDepth'] = data['WaterDepth'].round(2)
data['AfeCost'] = data['AfeCost'].round(2)
data['AfeDays'] = data['AfeDays'].round(2)
data['FinalCost'] = data['FinalCost'].round(2)
data['FinalDays'] = data['FinalDays'].round(2)
data['WellNptPercentageWow'] = data['WellNptPercentageWow'].round(2)
data['WellNptPercentage'] = data['WellNptPercentage'].round(2)
data['CompletionCostPlan'] = data['CompletionCostPlan'].round(2)
data['CompletionCostActual'] = data['CompletionCostActual'].round(2)
data['DrillingPlanWcpf'] = data['DrillingPlanWcpf'].round(2)
data['DrillingActualWcpf'] = data['DrillingActualWcpf'].round(2)
```

3. Standardize datetime values

```python
#Change to date only
date_cols = ['DocumentDate', 'SpudDate']

for col in date_cols:
    data[col] = pd.to_datetime(data[col], errors='coerce', utc=True)
    data[col] = data[col].dt.strftime('%Y-%m-%d')
```

```python
#Change to proper formate datetime
datetime_cols = ['SubmittedAt', 'WellStartDateTime', 'WellEndDateTime']

for col in datetime_cols:
    data[col] = (
        pd.to_datetime(data[col], errors='coerce', utc=True)
          .dt.strftime('%Y-%m-%d %H:%M:%S')
    )
```

Split the data according to the database schema

REGIONS

```python
regions = (
    data[['RegionName']]
    .drop_duplicates()
)

regions['RegionId'] = range(1, len(regions) + 1)

regions = regions[['RegionId', 'RegionName']]

regions
```

FIELDS

```python
fields = (
    data[['RegionName', 'FieldName']]
    .drop_duplicates()
)

fields = fields.merge(regions, on='RegionName', how='left')

fields['FieldId'] = range(1, len(fields) + 1)

fields = fields[['FieldId', 'RegionId', 'FieldName']]

fields
```

PAC

```python
pac = (
    data[['PacName']]
    .drop_duplicates()
)

pac = pac[pac['PacName'].notna()]

pac['PacId'] = range(1, len(pac) + 1)

pac = pac[['PacId', 'PacName']]

pac
```

WELLS

```python
wells = (
    data[['FieldName',
          'PacName',
          'WellName',
          'WellType',
          'WaterDepth',
          'Year']]
    .drop_duplicates()
)

wells = wells.merge(fields[['FieldId', 'FieldName']], on='FieldName', how='left')
wells = wells.merge(pac[['PacId', 'PacName']], on='PacName', how='left')

wells['WellId'] = range(1, len(wells) + 1)

wells = wells[['WellId', 'FieldId', 'PacId',
               'WellName', 'WellType', 'WaterDepth', 'Year']]

wells
```

RIGS

```python
rigs = (
    data[['RigName', 'RigType']]
    .drop_duplicates()
)

rigs['RigId'] = range(1, len(rigs) + 1)

rigs = rigs[['RigId', 'RigName', 'RigType']]

rigs
```

WELL_OPERATIONS

```python
operations = (
    data[
        [
            'WellName',
            'RigName',
            'AfeCost',
            'AfeDays',
            'SpudDate',
            'WellStartDateTime',
            'WellEndDateTime',
            'FinalCost',
            'FinalDays',
            'WellNptPercentageWow',
            'WellNptPercentage',
            'CompletionCostPlan',
            'CompletionCostActual',
            'DrillingPlanWcpf',
            'DrillingActualWcpf'
        ]
    ]
    .drop_duplicates()
)

operations = operations.merge(
    wells[['WellId', 'WellName']],
    on='WellName',
    how='left'
)

operations = operations.merge(
    rigs[['RigId', 'RigName']],
    on='RigName',
    how='left'
)

operations['OperationId'] = range(1, len(operations) + 1)

operations = operations[
    [
        'OperationId',
        'WellId',
        'RigId',
        'AfeCost',
        'AfeDays',
        'SpudDate',
        'WellStartDateTime',
        'WellEndDateTime',
        'FinalCost',
        'FinalDays',
        'WellNptPercentageWow',
        'WellNptPercentage',
        'CompletionCostPlan',
        'CompletionCostActual',
        'DrillingPlanWcpf',
        'DrillingActualWcpf'
    ]
]

operations
```

REPORTS

```python
reports = (
    data[['WellName',
          'RigName',
          'ReportType',
          'DocumentName',
          'DocumentDate',
          'SubmittedAt',
          'SubmittedBy']]
    .drop_duplicates()
)

reports = reports.merge(
    wells[['WellId', 'WellName']],
    on='WellName',
    how='left'
)

reports = reports.merge(
    rigs[['RigId', 'RigName']],
    on='RigName',
    how='left'
)

reports = reports.merge(
    operations[['OperationId', 'WellId', 'RigId']],
    on=['WellId', 'RigId'],
    how='left'
)

reports['ReportId'] = range(1, len(reports) + 1)

reports = reports[
    ['ReportId',
     'OperationId',
     'ReportType',
     'DocumentName',
     'DocumentDate',
     'SubmittedAt',
     'SubmittedBy']
]

reports
```

C) Load

Connect the database

```python
import sqlite3

conn = sqlite3.connect("drilling_operations.db")
```

Load the data according to the hierarchy

```python
# Parent tables first
regions.to_sql("REGIONS", conn, if_exists="append", index=False)
pac.to_sql("PAC", conn, if_exists="append", index=False)
```

```python
# Dependent tables
fields.to_sql("FIELDS", conn, if_exists="append", index=False)
wells.to_sql("WELLS", conn, if_exists="append", index=False)
rigs.to_sql("RIGS", conn, if_exists="append", index=False)
```

```python
# Fact tables
operations.to_sql("WELL_OPERATIONS", conn, if_exists="append", index=False)
reports.to_sql("REPORTS", conn, if_exists="append", index=False)
```

Close the connection

```python
conn.commit()
conn.close()
```
