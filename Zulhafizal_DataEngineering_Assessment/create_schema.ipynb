```python
"""
1. With `sqlite3`, a connection is created that allows us to execute SQL and interact with the database.

2. cursor.execute("PRAGMA foreign_keys = ON;") is being used in order to enable foreign key constraint.

3. All tables are created based on the ERD Design whereby the PAC, Regions, Rigs are dependent / parent table while Fields, Wells, Reports & Drilling_Operations are child table.

4. The relationship implemented as below : 
        PAC → Fields
        Regions → Fields
        Fields → Wells
        Wells → Reports
        Wells → Drilling_Operations
        Rigs → Drilling_Operations

5. Data type implemented as below : 
        INTEGER → IDs and numeric counters
        TEXT → Names and descriptions
        DECIMAL(p, s) → Precise numeric values (e.g., cost, depth)
        DATETIME → Time-related fields
"""

import sqlite3

db_name = "drilling_operations.db"

# Connect to database
conn = sqlite3.connect(db_name)
cursor = conn.cursor()

cursor.execute("PRAGMA foreign_keys = ON;")

# PAC table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS PAC (
        PACId INTEGER PRIMARY KEY AUTOINCREMENT,
        PACName TEXT NOT NULL
    );
''')

# Regions Table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS REGIONS (
        RegionId INTEGER PRIMARY KEY AUTOINCREMENT,
        RegionName TEXT NOT NULL UNIQUE
    );
''')

# Fields table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS FIELDS (
        FieldId INTEGER PRIMARY KEY AUTOINCREMENT,
        FieldName TEXT NOT NULL,
        PACId INTEGER NOT NULL,
        RegionId INTEGER NOT NULL,
        FOREIGN KEY (PACId) REFERENCES PAC(PACId),
        FOREIGN KEY (RegionId) REFERENCES REGIONS(RegionId)
        
    );
''')

# Rigs table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS RIGS (
        RigId INTEGER PRIMARY KEY AUTOINCREMENT,
        RigName TEXT NOT NULL,
        RigType TEXT
    );
''')

# Wells table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS WELLS (
        WellId INTEGER PRIMARY KEY AUTOINCREMENT,
        WellName TEXT NOT NULL,
        WellType TEXT,
        WaterDepth DECIMAL(10, 2),
        SpudDate DATETIME,
        WellStartDateTime DATETIME,
        WellEndDateTime DATETIME,
        FieldId INTEGER NOT NULL,
        FOREIGN KEY (FieldId) REFERENCES FIELDS(FieldId)
    );
''')

# Reports table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS REPORTS (
        ReportId INTEGER PRIMARY KEY AUTOINCREMENT,
        ReportType TEXT,
        DocumentName TEXT,
        DocumentDate DATETIME,
        SubmittedAt DATETIME,
        SubmittedBy TEXT,
        WellId INTEGER NOT NULL,
        FOREIGN KEY (WellId) REFERENCES WELLS(WellId)
    );
''')

# Drilling_Operation table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS DRILLING_OPERATIONS (
        DrillingId INTEGER PRIMARY KEY AUTOINCREMENT,
        Year INTEGER,
        AfeCost DECIMAL (15, 2),
        AfeDays DECIMAL (10, 2),
        FinalCost DECIMAL (15, 2),
        FinalDays DECIMAL (10, 2),
        WellNPTPercentageWOW DECIMAL (5, 2),
        WellNPTPercentage DECIMAL (5, 2),
        CompletionCostPlan DECIMAL (10, 2),
        CompletionCostActual DECIMAL (10, 2),
        DrillingplanWCPF DECIMAL (10, 2),
        DrillingActualWCPF DECIMAL (10, 2),
        WellId INTEGER NOT NULL,
        RigId INTEGER NOT NULL,
        FOREIGN KEY (WellId) REFERENCES WELLS(WellId),
        FOREIGN KEY (RigId) REFERENCES RIGS(RigId)

    );
''')

conn.commit()
conn.close()

print(f"Database '{db_name}' created successfully.")
```
